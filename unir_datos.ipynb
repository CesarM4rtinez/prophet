```python
import pandas as pd
import os

# Carga de los archivos
# Asegúrate de que los nombres de archivo coincidan con los que tienes en tu carpeta
file1 = 'funds_history-08.06.2026 (1).csv'
file2 = 'funds_history-08.06.2026.csv'

df1 = pd.read_csv(file1)
df2 = pd.read_csv(file2)

# Consolidación: Concatenar ambos DataFrames
df_consolidado = pd.concat([df1, df2], ignore_index=True)

# Limpieza: 
# 1. Convertir la columna de fecha a formato datetime para ordenarlas
df_consolidado['Modified (UTC)'] = pd.to_datetime(df_consolidado['Modified (UTC)'])

# 2. Ordenar cronológicamente (de la más antigua a la más reciente)
df_consolidado = df_consolidado.sort_values(by='Modified (UTC)', ascending=True)

# 3. Eliminar duplicados (si existieran registros idénticos en ambos archivos)
df_consolidado = df_consolidado.drop_duplicates()

# Guardar el resultado en un nuevo archivo
df_consolidado.to_csv('datos_bitacora.csv', index=False)

print("✅ Consolidación exitosa.")
print(f"Total de registros: {len(df_consolidado)}")
print("Archivo guardado como: 'datos_bitacora.csv'")
```
