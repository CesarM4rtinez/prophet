```python
import websocket
import json

def on_message(ws, message):
    data = json.loads(message)
    print("¡Datos recibidos!")
    print(data) # Imprime el JSON bruto para ver qué llega

def on_error(ws, error):
    print(f"Error: {error}")

def on_open(ws):
    print("Conexión establecida con éxito.")

# Ejecución directa sin hilos
ws = websocket.WebSocketApp(
    "wss://stream.binance.com:9443/ws/btcusdt@depth10@100ms",
    on_message=on_message,
    on_error=on_error,
    on_open=on_open
)

print("Intentando conectar...")
ws.run_forever()
```

```python
import requests
import matplotlib.pyplot as plt

url = "https://api.binance.com/api/v3/depth"
params = {"symbol":"BTCUSDT","limit":100}
data = requests.get(url, params=params).json()

bids = [(float(p), float(q)) for p,q in data['bids']]
asks = [(float(p), float(q)) for p,q in data['asks']]

bid_prices, bid_qty = zip(*bids)
ask_prices, ask_qty = zip(*asks)

plt.plot(bid_prices, bid_qty, color='green', label='Bids')
plt.plot(ask_prices, ask_qty, color='red', label='Asks')
plt.title("Gráfico de profundidad DOM BTC/USDT")
plt.xlabel("Precio")
plt.ylabel("Cantidad")
plt.legend()
plt.show()
```

```python
def on_message(ws, message):
    trade = json.loads(message)
    print(f"Trade ejecutado: Precio={trade['p']} Cantidad={trade['q']}")

ws = websocket.WebSocketApp("wss://stream.binance.com:9443/ws/btcusdt@trade",
                            on_message=on_message)
ws.run_forever()
```
